```python
import matplotlib.pyplot as plt
import numpy as np

# Confusion Matrix values
TP = 120
FN = 30
FP = 20
TN = 230

# Compute FPR
fpr = FP / (FP + TN)

print("False Positive Rate (FPR):", round(fpr, 4))
print("FPR Percentage:", round(fpr * 100, 2), "%")

# Confusion matrix
cm = np.array([[TP, FN],
               [FP, TN]])

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm)

ax.set_xticks([0, 1])
ax.set_yticks([0, 1])
ax.set_xticklabels(["Predicted Spam", "Predicted Not Spam"])
ax.set_yticklabels(["Actual Spam", "Actual Not Spam"])

for i in range(2):
    for j in range(2):
        ax.text(j, i, cm[i, j], ha="center", va="center", fontsize=12)

ax.set_title("Confusion Matrix - Email Spam Detection")
plt.colorbar(im)
plt.show()

# Bar chart for FPR components
labels = ["False Positives (FP)", "True Negatives (TN)"]
values = [FP, TN]

plt.figure(figsize=(6, 4))
plt.bar(labels, values)
plt.title("Values Used in False Positive Rate")
plt.ylabel("Count")
plt.show()

print(f"FPR = {FP} / ({FP} + {TN}) = {fpr:.4f} = {fpr*100:.2f}%")
print(f"Interpretation: {fpr*100:.2f}% of actual non-spam emails were wrongly marked as spam.")
```

```python
import matplotlib.pyplot as plt

# Problem 1 confusion matrix values
TP = 120
FN = 30
FP = 20
TN = 230

# Metrics
accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
sensitivity = recall
f1_score = 2 * (precision * recall) / (precision + recall)

# Print results
print("Problem 1 Metrics")
print(f"Accuracy    : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision   : {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall      : {recall:.4f} ({recall*100:.2f}%)")
print(f"Sensitivity : {sensitivity:.4f} ({sensitivity*100:.2f}%)")
print(f"F1 Score    : {f1_score:.4f} ({f1_score*100:.2f}%)")

# Data for graph
metrics = ["Accuracy", "Precision", "Recall", "Sensitivity", "F1 Score"]
values = [accuracy, precision, recall, sensitivity, f1_score]

# Bar chart
plt.figure(figsize=(8, 5))
bars = plt.bar(metrics, values)

plt.ylim(0, 1)
plt.ylabel("Score")
plt.title("Problem 1 Classification Metrics")

# Add values on top of bars
for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 0.02,
        f"{value:.2f}",
        ha='center'
    )

plt.show()
```

```python
import numpy as np

# Multi-class confusion matrix
cm = np.array([
    [30, 10, 5],
    [8, 40, 2],
    [3, 7, 25]
])

# Compute overall accuracy
correct_predictions = np.trace(cm)   # sum of diagonal
total_predictions = np.sum(cm)
accuracy = correct_predictions / total_predictions

print("Confusion Matrix:")
print(cm)

print("\nCorrect Predictions (Diagonal Sum):", correct_predictions)
print("Total Predictions:", total_predictions)
print(f"Overall Accuracy = {accuracy:.4f} ({accuracy*100:.2f}%)")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Multi-class confusion matrix
cm = np.array([
    [30, 10, 5],
    [8, 40, 2],
    [3, 7, 25]
])

# Compute accuracy
correct_predictions = np.trace(cm)
total_predictions = np.sum(cm)
accuracy = correct_predictions / total_predictions

print(f"Overall Accuracy = {accuracy:.4f} ({accuracy*100:.2f}%)")

# Plot confusion matrix
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(cm)

ax.set_xticks([0, 1, 2])
ax.set_yticks([0, 1, 2])
ax.set_xticklabels(["A", "B", "C"])
ax.set_yticklabels(["A", "B", "C"])

ax.set_xlabel("Predicted Class")
ax.set_ylabel("Actual Class")
ax.set_title("Confusion Matrix - Multi-Class Classification")

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(j, i, cm[i, j], ha="center", va="center", fontsize=12)

plt.colorbar(im)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Model A values
TP_A = 60
FN_A = 40
FP_A = 20
TN_A = 80

# Model B values
TP_B = 80
FN_B = 20
FP_B = 50
TN_B = 50

# Metrics
recall_A = TP_A / (TP_A + FN_A)
precision_A = TP_A / (TP_A + FP_A)

recall_B = TP_B / (TP_B + FN_B)
precision_B = TP_B / (TP_B + FP_B)

print("Model A")
print(f"Recall    = {recall_A:.4f} ({recall_A*100:.2f}%)")
print(f"Precision = {precision_A:.4f} ({precision_A*100:.2f}%)\n")

print("Model B")
print(f"Recall    = {recall_B:.4f} ({recall_B*100:.2f}%)")
print(f"Precision = {precision_B:.4f} ({precision_B*100:.2f}%)\n")

if recall_B > recall_A:
    print("Higher Recall: Model B")
else:
    print("Higher Recall: Model A")

if precision_A > precision_B:
    print("Higher Precision: Model A")
else:
    print("Higher Precision: Model B")
```
